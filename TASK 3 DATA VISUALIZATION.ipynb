```python
import pandas as pd

file_path = r"D:\data anaylaic\task 3\paint company.csv"

df = pd.read_csv(file_path)

print(df.head(20))
```

```python
date_column = "Year/Month"

df["Date"] = pd.NaT

# Monthly dates: 2020 Dec, 2021 Jan, etc.
monthly_mask = df[date_column].str.match(
    r"^\d{4}\s+[A-Za-z]{3}$",
    na=False
)

df.loc[monthly_mask, "Date"] = pd.to_datetime(
    df.loc[monthly_mask, date_column],
    format="%Y %b",
    errors="coerce"
)

# Financial years: 2016-17, 2017-18, etc.
financial_mask = df[date_column].str.match(
    r"^\d{4}-\d{2}$",
    na=False
)

df.loc[financial_mask, "Date"] = pd.to_datetime(
    df.loc[financial_mask, date_column].str[:4] + "-04-01",
    format="%Y-%m-%d",
    errors="coerce"
)

print(df[["Year/Month", "Date"]])
```

```python
print("\nDate Data Type:")
print(df["Date"].dtype)

print("\nMissing Dates:")
print(df["Date"].isna().sum())
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()

print(df.head(10))
```
